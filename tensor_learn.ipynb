```python
import torch

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

torch.manual_seed(42)
m1 = torch.rand(7,7).to(device)
m2 = torch.rand(1,7).to(device)

print(torch.matmul(m1, m2.T))
print(m1.device)

# 1. Set seeds for both CPU and GPU explicitly
torch.manual_seed(42)
torch.cuda.manual_seed(42)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# 2. Pass 'device=' inside rand() so it generates directly on the GPU
m1 = torch.rand(7, 7, device=device)
m2 = torch.rand(1, 7, device=device)

print(torch.matmul(m1, m2.T))
print(m1.device)
```

```python
import torch

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
torch.manual_seed(1234)
m1 = torch.rand(2, 3).to(device)
m2 = torch.rand(2, 3).to(device)

print(m1)
print(m2)

mul = torch.matmul(m1,m2.T)
max = torch.max(mul)
min = torch.min(mul)

print(mul)
print(max)
print(min)

max_idx = torch.argmax(mul)
min_idx = torch.argmin(mul)
print(max_idx)
print(min_idx)

torch.manual_seed(7)
t1 = torch.rand(1, 1, 1, 10)
print(t1, t1.shape)
t2 = torch.squeeze(t1)
print(t2, t2.shape)
```
